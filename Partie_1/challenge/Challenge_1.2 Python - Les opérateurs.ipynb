# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
def montant (prix, quantite, tva):
    return prix * quantite * (1 + tva)

prix = float(input("Entrez le prix: "))
quantite = int(input("Entrez la quantité: "))
tva = float(input("Entrez le taux de TVA (en pourcentage): ")) / 100

montant_final= montant(prix, quantite, tva)
print("Le montant final est de:", montant_final)

```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
duree_secondes = int(input("Entrez la durée en secondes: "))
heures = duree_secondes // 3600
minutes = (duree_secondes % 3600) // 60
secondes = duree_secondes % 60

print("Durée: {} heures, {} minutes, {} secondes".format(heures, minutes, secondes))
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
duree_heures = int(input("Entrez la durée en heures: "))
duree_minutes = int(input("Entrez la durée en minutes: "))
duree_secondes0 = int(input("Entrez la durée en secondes: "))
duree_secondes = duree_heures * 60 * 60 + duree_minutes * 60 + duree_secondes0

durée_definitive = "{} heures, {} minutes, {} secondes".format(duree_heures, duree_minutes, duree_secondes0)
print("Durée: {}".format(durée_definitive))

print("duree_secondes: {}".format(duree_secondes))

```
