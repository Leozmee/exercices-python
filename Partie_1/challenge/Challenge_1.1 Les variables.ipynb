Convertissez la variable `price` en int

```python
price = 5.4

price = int(price)
```

Que pouvez-vous remarquer ?

Réponse: int() supprime la partie décimale : 5.4 devient 5. Ce n'est pas un arrondi, la décimale est simplement coupée.

Trouvez le moyen d'arrondir la variable `temperature`.

```python
temperature = 5.3

temperature = round(temperature)
```

Après avoir arrondi la variable `temperature`, est-ce que c'est toujours un `float` ?

Réponse: Non, round() sans deuxième argument renvoie un int.

Ci-dessous, est-ce que height est un `int` ?

```python
height = 6.0

type(height).__name__
```

Réponse: Non, height est un float : 6.0 a une partie décimale, même si elle vaut 0.
