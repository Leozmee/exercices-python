- Which actor is mentioned the most often?


```python
import pandas as pd

link = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
df = pd.read_csv(link)
df.head()
```

```python
import ast

df['actors'] = df['actors_list'].apply(ast.literal_eval)

acteurs = df['actors'].explode()

print("Acteur le plus cité :", acteurs.value_counts().idxmax())
acteurs.value_counts().head(5)
```
